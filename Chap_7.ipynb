```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

# CHAP7 is a headerless, comma-separated file with two numeric columns.
data_path = Path("CHAP7")
data = np.loadtxt(data_path, delimiter=",", usecols=(0, 1))
x = data[:, 0]
y = data[:, 1]

plt.figure(figsize=(10, 5))
plt.plot(x, y, linewidth=1.5)
plt.xlabel("Column 1")
plt.ylabel("Column 2")
plt.title("CHAP7 data")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Loaded {len(data)} rows from {data_path.resolve()}")
```
